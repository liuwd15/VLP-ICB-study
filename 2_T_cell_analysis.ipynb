# proportion analysis

```r
tc_anno_percent_df <- tc@meta.data %>% group_by(annotation, treatment, sample, VLP, aCTLA4, batch) %>% summarise(count=n()) %>% 
    ungroup() %>% group_by(sample) %>% mutate(count_sample = sum(count)) %>% mutate(percentage = count/count_sample*100)
tc_anno_percent_df %>% select('sample','batch','VLP','aCTLA4','annotation','count') %>% 
    dcast(sample + batch + VLP + aCTLA4 ~ annotation) %>% mutate_all(~tidyr::replace_na(., 0)) %>% 
    write.csv('csv/b16_tc_annotation_cell_count.csv', row.names=F)
ggplot(tc_anno_percent_df, aes(x=treatment, y=percentage, color=treatment)) + geom_boxplot(color='gray', alpha=0.5, outliers=F) + 
    facet_wrap(~annotation, scales='free_y') + 
    theme_cowplot() + labs(x='',fill='treatment') + geom_jitter() +
    scale_color_brewer(palette='Paired') + labs(color='Treatment') +
    theme(legend.position=c(0.7,0.1), axis.text.x=element_blank())
ggsave('plot/b16_t_cluster_prop.pdf', width=7, height=7, device=cairo_pdf)
```

```r
tc_anno_percent_df$treatment <- factor(tc_anno_percent_df$treatment, levels=c('Control','αCTLA4 IV','αCTLA4 ID',
                                                                              'VLP','αCTLA4 IV + VLP','αCTLA4 ID + VLP'))
percent_fc <- tc_anno_percent_df %>% split(.$sample)  %>% 
    lapply(function(x) data.frame(CD4fc=log2(x$count[x$annotation=='CD4+Th1'] / x$count[x$annotation=='CD4+Treg']),
           CD8fc1=log2(x$count[x$annotation=='CD8+Te'] / x$count[x$annotation=='CD8+Tex']),
           treatment=unique(x$treatment)                      )) %>% bind_rows(.id = 'sample') %>%melt() 
stat_test <- percent_fc %>%
  rename(obs_type = variable) %>%   # <-- Rename "variable" to avoid the error
  group_by(obs_type) %>%
  t_test(value ~ treatment) %>%
  add_significance() %>%
  filter(p < 0.05) %>%   
  add_xy_position(x = "treatment", step.increase = 0.07) %>%
  rename(variable = obs_type) %>%
  as.data.frame()
percent_fc %>% 
  ggplot(aes(y = value, x = treatment, color = treatment)) + 
  geom_jitter() + 
  geom_boxplot(aes(color = treatment), alpha = 0.5, outlier.shape = NA) + 
  facet_wrap(~variable, scales = 'free_y', ncol = 2, 
             labeller = as_labeller(c('CD4fc' = 'CD4+Th1/Treg', 
                                     'CD8fc1' = 'CD8+Te/Tex'))) + 
  theme_cowplot() + 
  labs(x = 'Treatment', y = 'Log2 ratio', fill = 'treatment', color = 'Treatment') + 
  scale_color_manual(values = c('black', 'red', 'blue', '#F74ED6', '#FF8000', '#00C000')) +
  # Use stat_pvalue_manual instead of stat_compare_means
  stat_pvalue_manual(
    stat_test, 
    xmin = "group1",
    xmax = "group2",
    y.position = "y.position",
    label = "p", 
    tip.length = 0.01, 
    vjust = 0.2,
    inherit.aes = FALSE
  ) +
  theme(legend.position = 'none', 
        axis.text.x = element_text(angle = 45, vjust = 1, hjust = 1))
ggsave('plot/b16_box_cluster_ratio.pdf', width=6, height=5, unit='in', device=cairo_pdf)
```

# Differential expression

### comparison between treatments

```r
deg_list_combi_effect <- list()
Idents(tc) <- 'annotation'
annotation_i <- list('Tex'=c('CD8+Tex','CD8+Prolif'), 'Th1'='CD4+Th1', 'Te'='CD8+Te', 'Treg'='CD4+Treg','Prolif'='CD8+Prolif',
                     'ISG+T'='CD4/8+ISG+','HSP+T'='CD4/8+HSP+',
                     'Cd8T'=c('CD8+NK-like','CD8+Te','CD8+Prolif','CD8+Tex','CD8+Tn'),
                     'Cd4Th'=c('CD4+Th1','CD4+Tn','CD4+Th2','CD4+Th17'))
for(type_i in names(annotation_i))
{
    print(type_i)
    flush.console()
    suppressMessages({
        deg_list_combi_effect[[paste(type_i,'aCTLA4 ID + VLP vs VLP')]] <- FindMarkers(subset(tc, VLP=='ID'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='aCTLA4', ident.1='ID', ident.2='none', logfc.threshold=0.01, min.pct=0.1) 
        deg_list_combi_effect[[paste(type_i,'aCTLA4 IV + VLP vs VLP')]] <- FindMarkers(subset(tc, VLP=='ID'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='aCTLA4', ident.1='IV', ident.2='none', logfc.threshold=0.01, min.pct=0.1) 
        deg_list_combi_effect[[paste(type_i,'aCTLA4 ID + VLP vs aCTLA4 IV + VLP')]] <- FindMarkers(subset(tc, VLP=='ID'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='aCTLA4', ident.1='ID', ident.2='IV', logfc.threshold=0.01, min.pct=0.1) 
        deg_list_combi_effect[[paste(type_i,'aCTLA4 ID vs aCTLA4 IV')]] <- FindMarkers(subset(tc, VLP=='none'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='aCTLA4', ident.1='ID', ident.2='IV', logfc.threshold=0.01, min.pct=0.1)
        deg_list_combi_effect[[paste(type_i,'aCTLA4 ID + VLP vs aCTLA4 ID')]] <- FindMarkers(subset(tc, aCTLA4=='ID'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='VLP', ident.1='ID', ident.2='none', logfc.threshold=0.01, min.pct=0.1) 
        deg_list_combi_effect[[paste(type_i,'aCTLA4 IV + VLP vs aCTLA4 IV')]] <- FindMarkers(subset(tc, aCTLA4=='IV'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='VLP', ident.1='ID', ident.2='none', logfc.threshold=0.01, min.pct=0.1) 
        deg_list_combi_effect[[paste(type_i,'aCTLA4 ID vs Control')]] <- FindMarkers(subset(tc, VLP=='none'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='aCTLA4', ident.1='ID', ident.2='none', logfc.threshold=0.01, min.pct=0.1) 
        deg_list_combi_effect[[paste(type_i,'aCTLA4 IV vs Control')]] <- FindMarkers(subset(tc, VLP=='none'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='aCTLA4', ident.1='IV', ident.2='none', logfc.threshold=0.01, min.pct=0.1) 
        deg_list_combi_effect[[paste(type_i,'VLP vs Control')]] <- FindMarkers(subset(tc, aCTLA4=='none'), 
                           subset.ident=annotation_i[[type_i]], test.use='MAST', latent.vars='batch', 
                           group.by='VLP', ident.1='ID', ident.2='none', logfc.threshold=0.01, min.pct=0.1) 
    })
}
saveRDS(deg_list_combi_effect, 'data/b16_tc_deg_list_treatments_mast_batch.rds')
```

```r
volc_list <- list()
for(comparison_i in names(deg_list_combi_effect))
{
    df_i <- deg_list_combi_effect[[comparison_i]] %>% rownames_to_column('gene') %>%
                         filter(abs(avg_log2FC)<5.1)
    title_i <- comparison_i %>% str_replace('a','α') %>% str_replace('Cd8','CD8+') %>% str_replace('Cd4','CD4+')
    volc_list[[comparison_i]] <- ggplot(df_i, aes(x=avg_log2FC, y=-log10(p_val_adj))) + geom_point(aes(color=avg_log2FC)) +
            geom_hline(yintercept=-log10(0.05), linetype="dashed", color = "gray40") +
            geom_vline(xintercept=c(-0.5,0.5), linetype="dashed", color = "gray40") +
            geom_text_repel(data=df_i %>% filter(p_val_adj<0.05 & abs(avg_log2FC)>0.5) , 
                    aes(label=gene), max.overlaps=10) +
            theme_cowplot() + theme(legend.position='none') + 
            scale_color_gradient2(low='blue',mid='gray',high='red') + 
            labs(x=paste('Log2FC'), y='-log10(p.adj)', title=title_i) +
            theme(plot.title=element_text(face='plain',size=18), text=element_text(size=16))
}
```

```r
volc_group <- names(volc_list)
for(cell_i in names(annotation_i))
{
    volc_cell_i <- wrap_plots(volc_list[str_detect(volc_group, sub('\\+T','', cell_i))], ncol=3)
    ggsave(paste0('plot/b16_volcano_treatments',cell_i,'.pdf'), width=18, height=18, device=cairo_pdf)
}
```
